```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (10,6)
```

# Load Dataset

```python
df = pd.read_csv("../data/raw/ecommerce_data.csv")

df.head()
```

# Dataset Information


```python
df.info()
```

```python
df.describe(include="all")
```

# Missing Values

```python
missing = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Percentage": (df.isnull().sum()/len(df)*100).round(2)
})

missing
```

```python
plt.figure(figsize=(8,4))

sns.heatmap(df.isnull(), cbar=False)

plt.title("Missing Value Heatmap")

plt.show()
```

# Duplicate Records

```python
df.duplicated().sum()
```

# Outlier Detection

```python
columns = [
    "Quantity",
    "UnitPrice",
    "ItemsInCart",
    "TotalPrice"
]

for column in columns:

    plt.figure()

    sns.boxplot(x=df[column])

    plt.title(column)

    plt.show()
```

# Feature Engineering

```python
df["Date"] = pd.to_datetime(df["Date"])

df["Month"] = df["Date"].dt.month_name()

df["DayOfWeek"] = df["Date"].dt.day_name()

df["Quarter"] = "Q" + df["Date"].dt.quarter.astype(str)

df["AverageItemPrice"] = (
    df["TotalPrice"]/df["Quantity"]
)

df["DiscountApplied"] = np.where(
    df["CouponCode"].isna(),
    "No",
    "Yes"
)

median = df["TotalPrice"].median()

df["HighValueOrder"] = np.where(
    df["TotalPrice"]>=median,
    "Yes",
    "No"
)

df.head()
```

# Revenue by Product

```python
df.groupby("Product")["TotalPrice"]\
.sum()\
.sort_values(ascending=False)\
.plot(kind="bar")

plt.ylabel("Revenue")

plt.show()
```

```python
sns.countplot(data=df,x="PaymentMethod")

plt.show()
```

```python
sns.countplot(data=df,x="OrderStatus")

plt.show()
```

```python
corr = df.select_dtypes(include="number").corr()

sns.heatmap(corr,annot=True,cmap="Blues")

plt.show()
```

# Business Insights

```python
print("Total Revenue :",df["TotalPrice"].sum())

print("Average Order :",df["TotalPrice"].mean())

print("Top Product :",df["Product"].value_counts().idxmax())

print("Most Used Payment :",df["PaymentMethod"].value_counts().idxmax())

print("Top Referral :",df["ReferralSource"].value_counts().idxmax())
```
